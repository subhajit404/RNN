```python
quotes = quotes.str.replace(r'[^\w\s]', '', regex=True)
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000
tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
quotes.str.split().explode().unique().size
```

```python
word_idx = tokenizer.word_index
print(len(word_idx))
list(word_idx.items())[:10]
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
```

```python
for i in range (3):
    print(quotes[i])
```

```python
for i in range (3):
    print(sequence[i])
```

```python
X = []
y = []


for seq in sequence:
    for i in range(1,len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]
        X.append(input_seq)
        y.append(output_seq)

```
